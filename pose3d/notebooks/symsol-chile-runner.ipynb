```python
# ---- EXPERIMENT CONFIG: edit this cell for each run --------------------------
# SYMSOL-CHILE (pose3d/pose3d/datasets/symsol_chile.py; report: pose3d/reports/symmetry_spectrum.md): SYMSOL-style
# renders of regular pyramids with 7, 10 and 13 base sides. An n-gonal pyramid looks the same after a turn of 360/n
# degrees about its axis, so its pose density needs the Wigner-D degree l = n; a model that cannot reach it gets the
# axis right and guesses the turn, a median error of about 90/n degrees (12.9 / 9.0 / 6.9 deg).
# The data is RENDERED IN THIS KERNEL before training (seeded, so every kernel gets the same images; a few minutes on
# a GPU): 4,500 training views per shape (the 10 % setting of the symsol1_10pct_* runs) and the first 1,000 test views
# per shape, which are the ones scored. Nothing to attach for the data.
#
# RUN ONE VARIANT PER KERNEL (set RUN), as the other runners (gpu_pool/rtx_monitor.py keys its state by kernel):
#   i2s_l6     Image2Sphere (--model i2s_real), Wigner-D degrees up to L = 6, AdamW recipe of i2s-symsol-runner.
#              Prediction: below every pyramid's degree n, so the turn is a guess (reports/symmetry_spectrum.md).
#   i2s_l10    the same Image2Sphere with --lmax 10 (1,771 Wigner coefficients; the level-5 evaluation grid's Wigner matrix
#              is ~17 GB, fine on an RTX Pro 6000). Reaches the 7- and 10-gon's degree, not the 13-gon's: the turn of the
#              first two can be represented, the 13-gon's still cannot.
#   stf_l10    the Clifford Flow with the STF tensor track up to rank 10 (--stf_track --stf_lmax 10), the recipe of
#              symsol-stf-runner. Ranks 7 and 10 reach the 7- and 10-gon's degree, rank 13 is beyond it.
# Scores: acc@15 / median error to the nearest equivalent pose per shape (and I2S's grid log-likelihood).
# Estimated wall time on one RTX Pro 6000 (from the measured SYMSOL I runs, scaled to 13,500 training images):
# i2s_l6 ~25 min, stf_l10 ~45-75 min (rank 10 costs ~70x rank 6 inside the STF track); on a T4 ~2 h / ~4-6 h.
# Over the 30-minute limit of CLAUDE.md: a human approves each launch.
#
# OFFLINE (RTX Pro 6000 has no internet) attach the repo snapshot of this branch and the offline-deps datasets listed
# in symsol-stf-runner.ipynb (no SYMSOL dataset needed). ONLINE (T4): the branch is cloned and the dependencies
# installed from PyPI / GitHub.
RUN = 'i2s_l6'
WANDB_API_KEY = None
REPO = 'chagrygoris/GA-Research'
BRANCH = 'claude/symsol-symsol2-differences-2dukos'
DATA_ARGS = ['--dataset', 'symsol', '--symsol_set', '5', '--symsol_train_views', '4500', '--symsol_test_views', '1000',
             '--no-use_warp', '--no-use_synth', '--no-ema', '--no-ddp', '--batch_size', '64', '--n_epochs', '40',
             '--num_workers', '4', '--encoder', 'resnet50']
ADAMW = ['--optimizer', 'adamw', '--lr', '1e-4', '--lr_scaling', 'sqrt', '--lr_schedule', 'cosine', '--warmup_epochs', '5']
VARIANTS = {
    'chile_10pct_i2s_l6_adamw': ['--model', 'i2s_real', '--lmax', '6', '--no-fixed_val_noise', '--i2s_eval_rec_level', '5',
                                 *ADAMW],
    'chile_10pct_i2s_l10_adamw': ['--model', 'i2s_real', '--lmax', '10', '--no-fixed_val_noise', '--i2s_eval_rec_level', '5',
                                  *ADAMW],
    'chile_10pct_gatr_stf_l10': ['--model', 'clifford_flow', '--fixed_val_noise', '--no-symsol_likelihood', *ADAMW,
                                 '--stf_track', '--stf_lmax', '10'],
}
SHORT = {'i2s_l6': 'chile_10pct_i2s_l6_adamw', 'i2s_l10': 'chile_10pct_i2s_l10_adamw', 'stf_l10': 'chile_10pct_gatr_stf_l10'}
# (RTX, other GPU) minutes. A kill cannot be resumed (the checkpoint is written only after the last epoch), so the
# limits are generous: ~3x the estimate for stf_l10, whose rank-10 cost on a GPU is not measured.
WATCHDOG_MINUTES = {'i2s_l6': (60, 240), 'i2s_l10': (120, 480), 'stf_l10': (180, 600)}
assert RUN in SHORT, f'RUN must be one of {list(SHORT)}'
# a repo snapshot / clone must carry SYMSOL-CHILE and rank-10 STF tensors
MARKERS = {'pose3d/pose3d/datasets/symsol.py': 'symsol_chile', 'pose3d/pose3d/models/stf_track.py': 'MAX_RANK = 10',
           'pose3d/pose3d/engine/resume.py': 'RESUMABLE_EXIT'}
```

```python
os.chdir('/kaggle/working/GA-Research/pose3d')   # the package is found from here, no install needed
DATA = '/kaggle/working/data'
# Render SYMSOL-CHILE (seeded: the same images in every kernel). Skipped when an earlier cell run already did it.
if not os.path.exists(f'{DATA}/symsol_chile/train/rotations.npz'):
    subprocess.run([sys.executable, '-u', '-m', 'pose3d.datasets.symsol_chile', f'--root={DATA}', '--train', '4500',
                    '--test', '1000', '--device', 'cuda'], check=True)
print('SYMSOL-CHILE root:', DATA, flush=True)
name = SHORT[RUN]
# RESUME (engine/resume.py): the run writes /kaggle/working/<run_name>_resume.pth after every epoch, which Kaggle keeps
# as this kernel's output. To continue a stopped run, launch a new kernel with the stopped one attached as input
# (gpu_pool/launcher.py --kernel-source <owner>/<kernel>): its checkpoint is found here and passed as --resume.
resume = sorted(glob.glob(f'/kaggle/input/**/{name}_resume.pth', recursive=True))
resume_args = []
if resume:
    shutil.copy(resume[0], f'/kaggle/working/{name}_resume.pth')     # keep advancing the same file
    resume_args = [f'--resume=/kaggle/working/{name}_resume.pth']
    print('resuming from', resume[0], flush=True)
cmd = [sys.executable, '-u', '-m', 'pose3d', f'--path_to_datasets={DATA}', '--platform=kaggle', f'--run_name={name}',
       '--resume_dir=/kaggle/working', *resume_args, *DATA_ARGS, *VARIANTS[name]]
print(' '.join(cmd), flush=True)
# The output is also written to /kaggle/working/<run_name>.log: Kaggle keeps /kaggle/working as the kernel's output
# even when its own log stream comes back empty (seen on long RTX sessions), so the per-epoch metrics survive.
# WATCHDOG: the training process is killed after MAX_MINUTES (WATCHDOG_MINUTES, about twice the estimate) so
# that a stuck session cannot burn the 12-hour GPU budget (two sessions did exactly that).
gpu_name = subprocess.run(['nvidia-smi', '--query-gpu=name', '--format=csv,noheader'], capture_output=True, text=True).stdout.strip()
MAX_MINUTES = WATCHDOG_MINUTES[RUN][0 if 'RTX' in gpu_name else 1]
print(f'GPU: {gpu_name!r}, watchdog {MAX_MINUTES} min', flush=True)
import threading, time
with open(f'/kaggle/working/{name}.log', 'w') as fh:
    proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, env={**os.environ})
    def _watchdog():
        time.sleep(MAX_MINUTES * 60)
        if proc.poll() is None:
            # SIGTERM first: the run exits with 75 and keeps its last epoch's resume checkpoint; SIGKILL only if stuck.
            print(f'WATCHDOG: stopping the run after {MAX_MINUTES} minutes (resumable)', flush=True)
            proc.terminate()
            try:
                proc.wait(timeout=60)
            except subprocess.TimeoutExpired:
                proc.kill()
    threading.Thread(target=_watchdog, daemon=True).start()
    for line in proc.stdout:
        print(line, end='', flush=True)
        fh.write(line); fh.flush()
    rc = proc.wait()
# 75: stopped (watchdog / SIGTERM) with a resume checkpoint in /kaggle/working; the kernel ends cleanly so its output
# is kept, and a new kernel with this one attached continues it. Anything else non-zero is a crash and fails the kernel.
if rc == 75:
    print(f'STOPPED, RESUMABLE: /kaggle/working/{name}_resume.pth; relaunch with this kernel as --kernel-source', flush=True)
else:
    assert rc == 0, f'pose3d exited with code {rc}'
```
